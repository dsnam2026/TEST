## 1️⃣ Comments vs. Code

In Python, every line is either a **comment** or **code**.

- **Comment**: a line starting with `#`. Python **ignores it completely** — it never runs. It's a note for humans.
  (주석: 컴퓨터가 실행하지 않고 무시하는, 사람을 위한 메모)
- **Code**: a line that the computer actually **executes**.
  (코드: 컴퓨터가 실제로 실행하는 명령어)

Good comments explain *why* the code was written a certain way — very helpful for your future self, or teammates.

````python
# This line is a comment. It will not run.
print("This line is code. It runs!")  # you can also add a comment after code

# print("This line is commented out, so it will NOT run")
````

## 2️⃣ Input vs. Output

- **Output**: `print()` — shows something on the screen.
  (출력: 화면에 결과를 보여주는 것)
- **Input**: `input()` — asks the user to type something.
  (입력: 사용자로부터 값을 받는 것)

⚠️ **Important**: whatever `input()` receives is always a **string** — text made of characters, like `"apple"` or `"Hello!"`.
(string: 글자·숫자·기호가 나열된 텍스트 형태의 데이터)

````python
print("Hello! Welcome to English Class.")   # output

name = input("What is your name? ")          # input (user types their answer)
print("Nice to meet you,", name, "!")        # print the input back
````

## 4️⃣ Operators

Today, we'll focus on the **four basic arithmetic operators** (사칙연산):

| Operator | Meaning | Example |
|---|---|---|
| `+` | addition (덧셈) | `3 + 2` → `5` |
| `-` | subtraction (뺄셈) | `3 - 2` → `1` |
| `*` | multiplication (곱셈) | `3 * 2` → `6` |
| `/` | division (나눗셈) | `3 / 2` → `1.5` |

There are also **comparison operators**: `==` `!=` `>` `<` `>=` `<=` → these always give you `True` or `False`.
(비교 연산자: 두 값을 비교해서 True 또는 False를 알려줌)

````python
score = 17
total = 20

percentage = score / total * 100        # arithmetic operators
print(percentage)

is_pass = percentage >= 80              # comparison operator → True/False
print(is_pass)
````

## 5️⃣ Functions — Basic Idea

A function is code you can name and reuse. You define it with `def`, then "call" it by name whenever you need it.
(함수: 이름을 붙여서 필요할 때마다 다시 사용할 수 있는 코드 묶음)

```python
def function_name(parameter):
    code to run
    return result   # optional
```

`print()` and `input()` are actually functions that Python already built for you!

````python
def greet(name):
    return f"Hello, {name}! Nice to meet you."

message = greet("Amy")
print(message)

print(greet("Minsu"))   # reuse the same function with a different value
````

````python
def nemo(name):
    return f"Hello, {name}! Nice to meet you."

message = nemo(input ("What's your name?"))
print(message)

print(nemo("Bimo"))
````

### 🚀 Demo 3: Build an Answer-Checking Function
Once you turn "checking an answer" into a function, you can reuse it for many questions. This function will come back in our final demo!

````python
def check_answer(user_answer, correct_answer):
    # ignore uppercase/lowercase and extra spaces
    return user_answer.strip().lower() == correct_answer.strip().lower()

print(check_answer("Apple", "apple"))     # True
print(check_answer(" APPLE ", "apple"))   # True
print(check_answer("banana", "apple"))    # False
````

## 6️⃣ Module · Library · Package

- **Module**: a single Python file full of reusable code. You bring it in with `import module_name`.
  (모듈: 재사용 가능한 코드가 담긴 파일 하나)
- **Library**: a collection of modules. (`random` is a library already built into Python.)
  (라이브러리: 여러 모듈을 모아놓은 것)
- **Package**: a collection of libraries, organized and shared online. You install one with `pip install package_name`.
  (패키지: 라이브러리들을 체계적으로 묶어 배포하는 단위. 예: 영어 학습에 쓰이는 `gTTS`, `googletrans`)

Today we'll use `random`, which comes built into Python — no installation needed.

````python
import random

print(random.randint(1, 10))           # a random whole number from 1 to 10
print(random.choice(["A", "B", "C"]))  # randomly pick one item from a list
````

### 🎉 Final Demo: Random English Vocabulary Quiz
Now let's combine **everything** you learned today — comments, input/output, variables, operators, functions, and modules — into one real quiz app! Run it and try a few questions.

> 💡 This uses a `for` loop and list indexing, which you haven't learned yet. That's okay — you'll learn them next time!

````python
#@markdown ### 🧠 Random English Vocabulary Quiz (click Run!)
#@markdown Type the Korean meaning of the word shown, and get instant feedback.

import random

# word list and meaning list (we'll learn dictionaries next time - two lists work fine for now!)
word_list = ["apple", "book", "friend", "school", "happy"]
meaning_list = ["사과", "책", "친구", "학교", "행복한"]

def check_answer(user_answer, correct_answer):
    return user_answer.strip().lower() == correct_answer.strip().lower()

score = 0
total_questions = 3

for i in range(total_questions):
    index = random.randint(0, len(word_list) - 1)
    question_word = word_list[index]
    correct_meaning = meaning_list[index]

    user_answer = input(f"Q{i+1}. What does '{question_word}' mean? (type in Korean) ")

    if check_answer(user_answer, correct_meaning):
        print("✅ Correct!\n")
        score += 1
    else:
        print(f"❌ Not quite! The correct answer was '{correct_meaning}'.\n")

percentage = score / total_questions * 100
print(f"🏁 Final score: {score}/{total_questions} ({percentage:.0f}%)")
````
